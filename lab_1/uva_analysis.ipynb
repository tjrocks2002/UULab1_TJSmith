```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as scs

df = pd.read_csv('ForeignGifts_edu.csv').sort_values(by='Institution Name', ascending=False)
#df.head()
```

Compared to the overall market, UVA received a very small portion of available donations. UVA received the most funding from Bangaladesh, followed by Sweden, Tanzania, Switzerland, and England.\
\
This is interesting as only England falls within the top 5 gifting countries by amount. Switzerland does fall into the top 5 gifting countries by count.\
\
If we compare to a fellow institution like Virginia Commonwealth University, we can see that they receive a massive amount of donations in comparison (~90M compared to UVA's ~22M). The majority of their donations, by comparison, came from Qatar (from whom they received ~$79M, more than double UVA's total donations received). This clearly indicates that we need to change our strategy when soliciting gifts.




```python
alltotaldol = df.groupby('Institution Name')['Foreign Gift Amount'].sum()

#UVA Statistics:
UVAtotaldol = alltotaldol['University of Virginia']
UVApercentile = scs.percentileofscore(alltotaldol,UVAtotaldol,kind='mean')
UVApercentofall = UVAtotaldol/alltotaldol.sum()

print(f"UVA's total dollars: {UVAtotaldol:,}")
print(f"UVA's percentile was: {UVApercentile/100:.01%}")
print(f"UVA received {UVApercentofall:.0001%} of all donations in the market")

myct = pd.crosstab(df['Institution Name'],df['Country of Giftor'], values=df['Foreign Gift Amount'], aggfunc = 'sum', margins=True, margins_name='Total', normalize = False, dropna=True)
UVAct = myct.loc[['University of Virginia']].dropna(axis=1,how='any').stack()
print("\nUVA's Top Donations:")
print(UVAct.iloc[:-1].sort_values(ascending=False).astype(int))

#VCU Statistics
VCUtotaldol = alltotaldol['Virginia Commonwealth University']
VCUpercentile = scs.percentileofscore(alltotaldol,VCUtotaldol,kind='mean')
VCUpercentofall = VCUtotaldol/alltotaldol.sum()

print(f"VCU's total dollars: {VCUtotaldol:,}")
print(f"VCU's percentile was: {VCUpercentile/100:.01%}")
print(f"VCU received {VCUpercentofall:.0001%} of all donations in the market")

myct = pd.crosstab(df['Institution Name'],df['Country of Giftor'], values=df['Foreign Gift Amount'], aggfunc = 'sum', margins=True, margins_name='Total', normalize = False, dropna=True)
VCUct = myct.loc[['Virginia Commonwealth University']].dropna(axis=1,how='any').stack()
print("\nVCU's Top Donations:")
print(VCUct.iloc[:-1].sort_values(ascending=False).astype(int))
```

**The three categories we would want to consider when selecting a country to solicit gifts from would be:**
1. **Number of gifts donated on average:**
    1. From our initial analysis, we know that the number of gifts donated on average across all countries is **182**. So we would want to select countries who fall above this mean. These countries include:




```python
countrycounts = df.groupby('Country of Giftor')['Foreign Gift Amount'].count().sort_values(ascending=False)
abavgcount = countrycounts.loc[countrycounts > countrycounts.mean()]
print(abavgcount)
```

2. **The average size of those gifts:**
    1. From our initial analysis, we know that the average size of a gift donated across all countries is **$588,232.72**. We also want to select countries who fall above this mean. These countries include:

```python
countrygiftsize = df.groupby('Country of Giftor')['Foreign Gift Amount'].mean().sort_values(ascending=False)
abavggiftsize = countrygiftsize.loc[countrygiftsize > countrygiftsize.mean()]
print(abavggiftsize)
```

3. **The total amount of gifts (in dollars) the country will give:**
    1. Finally, we know from our initial analysis that the average total gift amount from all countries in the dataset is **$107,100,100.09**. We want countries who also donate above this amount. These countries include:

```python
countrytotgifts = df.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)
abavgtotgifts = countrytotgifts.loc[countrytotgifts > countrytotgifts.mean()]
print(abavgtotgifts)
```

**The following countries are found in every list:**

```python
countindex = set(abavgcount.index)
sizeindex = set(abavggiftsize.index)
totindex = set(abavgtotgifts.index )
potentialdonors = countindex & sizeindex & totindex
print(potentialdonors)
```

**We can compare these countries by the three attributes discussed to help in choosing which to solicit futher donations from, and add an additional consideration of how many independent organizations there are to solicit from in each country.**

```python
comparisontable = df.groupby('Country of Giftor')['Foreign Gift Amount'].agg(['count', 'mean', 'sum']).sort_values(by='sum', ascending=False)
fullcomparetable = comparisontable.loc[comparisontable.index.isin(potentialdonors)]
numberoforgs = df.groupby('Country of Giftor')['Giftor Name'].nunique().sort_values(ascending=False)
fullcomparetable['number_of_orgs'] = numberoforgs[fullcomparetable.index]
print(fullcomparetable)

```
